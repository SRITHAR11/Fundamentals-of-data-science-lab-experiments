```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Preprocessing Dataset.csv")
df
```

```python
df.info()
```

```python
df["Country"].mode()
```

```python
country_mode = df["Country"].mode()[0]

df["Country"] = df["Country"].fillna(country_mode)

df
```

```python
age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

df
```

```python
salary_mean = round(df["Salary"].mean())

df["Salary"] = df["Salary"].fillna(salary_mean)

df
```

```python
df.isnull().sum()
```

```python
country_dummies = pd.get_dummies(df["Country"])

country_dummies
```

```python
updated_dataset = pd.concat(
    [country_dummies, df[["Age", "Salary", "Purchased"]]],
    axis=1
)

updated_dataset
```

```python
updated_dataset["Purchased"] = updated_dataset["Purchased"].replace({
    "No": 0,
    "Yes": 1
})

updated_dataset
```
